#  Lists and Tuples – Understanding Structures and Their Functions


# Difference Between List and Tuple

| Feature        | List (Zepto Basket)                          | Tuple (Fixed Package)                  |
|----------------|----------------------------------------------|----------------------------------------|
| Mutability     | Mutable – items can be changed, added, removed | Immutable – items cannot be changed once created |
| Syntax         | Uses square brackets `[]`                      | Uses parentheses `()` or commas `,`       |
| Example        | `basket = ["milk", "bread", "eggs"]`           | `items = ("milk", "bread", "eggs")`       |
| Memory         | Consumes more memory (flexible structure)      | Consumes less memory (fixed structure)    |
| Use Cases      | Dynamic data – e.g., Zepto cart where you add/remove groceries | Fixed data – e.g., coordinates, dictionary keys, offers |
| Methods        | Many methods: `append()`, `extend()`, `insert()`, `remove()`, `pop()`, `sort()`, `reverse()` | Limited methods: `count()`, `index()`     |
| Performance    | Slower compared to tuples (because of mutability) | Faster for iteration and fixed data       |

---

## List Example
```python
basket = ["milk", "bread", "eggs"]
basket.append("butter")   # Add new item
print(basket)
# Output: ['milk', 'bread', 'eggs', 'butter']


## Tuple Example
```python
items = ("milk", "bread", "eggs")

print(items)
# Output: ('milk', 'bread', 'eggs')

# Trying to add a new item like in a list:
# items.append("butter")   ❌ Error (tuples are immutable)

# Correct way: create a new tuple by combining
new_items = items + ("butter",)
print(new_items)
# Output: ('milk', 'bread', 'eggs', 'butter')



1. Mutability:

- Lists → Mutable (you can change, add, or remove items).
- Tuples → Immutable (once created, data cannot be changed).

2. Syntax

````python
basket = ['milk' , 'bread' , 'eggs']
````

````python
items = ('milk' , 'bread' , 'eggs')
````

````python
# Even tuple without parenthesis
a = 1, 2
````

````python
print(basket)
print(items)
print(a)
````

3. Memory Usage

- Lists consume more memory because they allow dynamic changes.
- Tuples consume less memory since they are fixed and lightweight.

4. Use Cases

Lists:

- Best for dynamic data that changes often.
- Example: Zepto basket where you keep adding/removing groceries.

````python
basket = ['Milk' ,'Bread' ]
print(basket)
````

````python
# Add new element into list
basket.append("Eggs")
````

````python
print(basket)
````

Tuples:

- Best for fixed data that should not change.
- Example: Coordinates, dictionary keys, or function arguments

````python
location = (18.5204, 73.8567)   # Pune coordinates
print(location)
````

### Methods for Lists and Tuples

# 🛒 List Methods with Zepto Basket Records

| Method    | What it does                        | Zepto Example (Basket)                          | Output Example |
|-----------|-------------------------------------|------------------------------------------------|----------------|
| append()  | Add one item at the end             | `basket.append("butter")`                      | ['milk','bread','eggs','butter'] |
| extend()  | Add multiple items                  | `basket.extend(["apple","banana"])`            | ['milk','bread','eggs','apple','banana'] |
| insert()  | Add at a specific position          | `basket.insert(1,"jam")`                       | ['milk','jam','bread','eggs'] |
| remove()  | Remove first matching item          | `basket.remove("milk")`                        | ['bread','eggs'] |
| pop()     | Remove by index (default last)      | `basket.pop()`                                 | removes last item → ['milk','bread'] |
| clear()   | Empty the list                      | `basket.clear()`                               | [] |
| index()   | Find position of item               | `basket.index("bread")`                        | 1 |
| count()   | Count how many times item appears   | `basket.count("apple")`                        | 2 |
| sort()    | Sort items alphabetically           | `basket.sort()`                                | ['apple','banana','bread','milk'] |
| reverse() | Reverse order                       | `basket.reverse()`                             | ['milk','bread','apple'] |


````python
# Example
basket = ["milk", "bread"]
basket.append("eggs")
print(basket)   # ['milk', 'bread', 'eggs']
````

### Tuple Methods (limited)

- count() → counts occurrences of a value
- index() → finds position of a value

````python
items = ("milk" , "bread" , "milk" )
````

````python
# Count
items.count('milk')
````

````python
items.count('bread')
````

````python
# Index
items.index('milk')
````

- Use lists when you need flexibility (like a Zepto cart).

- Use tuples when data must remain fixed (like GPS coordinates or config values).

# Python Operators Quick Reference

## 1. Comparison Operators
| Operator | Meaning                  | Example                | Output   |
|----------|--------------------------|------------------------|----------|
| `==`     | Equal to                 | `5 == 5`               | True     |
| `!=`     | Not equal to             | `5 != 3`               | True     |
| `>`      | Greater than             | `10 > 7`               | True     |
| `<`      | Less than                | `2 < 5`                | True     |
| `>=`     | Greater than or equal to | `5 >= 5`               | True     |
| `<=`     | Less than or equal to    | `3 <= 4`               | True     |

---

## 2. Arithmetic Operators
| Operator | Meaning        | Example       | Output |
|----------|----------------|---------------|--------|
| `+`      | Addition       | `5 + 3`       | 8      |
| `-`      | Subtraction    | `10 - 4`      | 6      |
| `*`      | Multiplication | `6 * 2`       | 12     |
| `/`      | Division       | `8 / 2`       | 4.0    |
| `//`     | Floor division | `9 // 2`      | 4      |
| `%`      | Modulus        | `9 % 2`       | 1      |
| `**`     | Exponentiation | `2 ** 3`      | 8      |

---

## 3. Logical Operators
| Operator | Meaning   | Example                  | Output |
|----------|-----------|--------------------------|--------|
| `and`    | Both true | `(5 > 3) and (2 < 4)`    | True   |
| `or`     | Either    | `(5 > 3) or (2 > 4)`     | True   |
| `not`    | Negation  | `not(5 > 3)`             | False  |

---

## 4. Assignment Operators
| Operator | Meaning                  | Example       | Output |
|----------|--------------------------|---------------|--------|
| `=`      | Assign value             | `x = 5`       | x = 5  |
| `+=`     | Add and assign           | `x = 5; x += 3` | x = 8 |
| `-=`     | Subtract and assign      | `x = 5; x -= 2` | x = 3 |
| `*=`     | Multiply and assign      | `x = 4; x *= 2` | x = 8 |
| `/=`     | Divide and assign        | `x = 8; x /= 2` | x = 4.0 |
| `//=`    | Floor divide and assign  | `x = 9; x //= 2` | x = 4 |
| `%=`     | Modulus and assign       | `x = 9; x %= 2` | x = 1 |
| `**=`    | Exponent and assign      | `x = 2; x **= 3` | x = 8 |

---

## 5. Membership Operators
| Operator | Meaning                   | Example                  | Output |
|----------|---------------------------|--------------------------|--------|
| `in`     | Value exists in sequence  | `"milk" in ["milk","eggs"]` | True |
| `not in` | Value not in sequence     | `"bread" not in ["milk","eggs"]` | True |

---

## 6. Identity Operators
| Operator | Meaning                  | Example           | Output |
|----------|--------------------------|-------------------|--------|
| `is`     | Same object              | `x is y`          | True/False |
| `is not` | Not same object          | `x is not y`      | True/False |


## If Else revise with List and Tuple

````python
# Using List
basket = ["milk" , "bread" ]
````

````python
# check if "chips" is in the list
if "chips" in basket:
  print("It's added already")
else:
  print("Add chips to basket")
````

````python
# Using Tuple
````

````python
items = ("milk" , "bread" , "eggs")
````

````python
if "butter" in items:
  print("Butter already in basket")
else:
  print("Butter not Found - add separately")
````

````python
# Add or remove item
basket = ['milk' , 'bread' , 'chips']
````

````python
if 'chips' in basket:
  basket.remove("chips")
  print("Removed chips from basket" , basket)
else:
  basket.append('chips')
  print("Added chips into basket" , basket)
````

````python
# Check quantity
basket = ['apple' ,'apple' , 'banana']
````

````python
if basket.count('apple') > 1:
  print("You have multiple apples in basket")
else:
  print("only one apple inside basket")
````
