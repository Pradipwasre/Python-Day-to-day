# 1. For Loops

- A for loop is used when you want to repeat an action a fixed number of times or iterate over items in a sequence (like a list).



````python
# Example – Printing "Hello" 100 times
````

````python
for i in range(100):
  print("hello")
````

````python
basket = ['Milk' , 'Bread' , "Eggs" , "Rice"]
````

````python
for i in basket:
  print(i)
````

````python
# PRINT 1 TO 5:
for i in range(1,6):
  print(i)
````

````python
 # Write a code print "Hey there" for 5 times.
 for i in range(5):
  print('hey there')
````

````python
# Print squares of number: Start = 1 , end = 10
for i in range(1,11):
  print(i ** 2)
````

````python
# Print Characters in a Word
for i in 'zepto':
  print(i)
````

````python
# Fixed secret number
secret_number = 97

guess = None

while guess != secret_number:
    guess = int(input("Guess a number between 0 and 100: "))

    if guess < secret_number:
        print("Too low! Try again.")
    elif guess > secret_number:
        print("Too high! Try again.")
    else:
        print("Correct! The secret number was", secret_number)
````

# What is a while loop?

- A while loop repeats a block of code as long as a condition is true.
It is useful when you don’t know in advance how many times you need to repeat.

- The loop runs until the condition becomes False.
- Be careful: if the condition never becomes False, the loop runs forever (infinite loop).

````python
i = 5
i < 5
````

````python
i = 0

while i < 5:
  print("hello")
  i += 1
````

# Controlling Loops in Python

## Introduction
In Python, loops normally run until their condition ends.  
Sometimes, we need **extra control** inside the loop.  
Python provides three special statements for this purpose:  
- `break` → Exit the loop completely.  
- `continue` → Skip the current iteration and move to the next.  
- `pass` → Do nothing (acts as a placeholder).  

---

## 1. break
### Theory
- The `break` statement is used to **exit the loop immediately**, even if the loop condition is still true.
- Useful when you find what you are looking for and don’t need to continue checking further.

### Example

```python
basket = ["milk", "bread", "chips", "eggs"]

for item in basket:
    if item == "chips":
        print("Found chips – stop checking!")
        break
    print("Checking item:", item)

---

## 2. continue

### Theory
The `continue` statement is used to **skip the current iteration** and move to the next one.  
It is useful when you want to ignore certain items but keep looping through the rest.

### Example
```python
basket = ["milk", "bread", "chips", "eggs"]

for item in basket:
    if item == "chips":
        continue   # skip chips
    print("Healthy item:", item)

## 3. pass

### Theory
- The `pass` statement is a **placeholder**.  
- It does nothing — the loop continues normally.  
- Useful when you want to write code later but keep the structure valid.

### Example
```python
basket = ["milk", "bread", "chips"]

for item in basket:
    if item == "chips":
        pass   # do nothing for now
    else:
        print("Item:", item)

# Example on `continue`

## Question
In Zepto, a basket has both healthy and junk items.  
Write a program using a `for loop` with `continue` to **skip junk items ("chips", "cola")** and only print the healthy items from the basket.

---

## Answer
```python
basket = ["milk", "bread", "chips", "eggs", "cola", "rice"]

    if item == "chips" or item == "cola":
        continue   # skip junk items
    print("Healthy item:", item)

## 2. Count Distinction Marks (Hawkins Students)

### Problem
Given marks of Hawkins students, print how many scored **75 or more** (Distinction).

### Solution
```python
marks = [85, 60, 92, 45, 77, 50, 90]

count = 0
for m in marks:
    if m >= 75:
        count += 1

print("Number of students with Distinction:", count)

## 4. Skip Monster (Using continue)

### Problem
Loop through a basket of items.  
If the item is **"Demogorgon"**, skip it.

### Solution
```python

### Problem
Loop through Stranger Things villains.  
Stop the loop once **"Vecna"** is found.

### Solution
```python
