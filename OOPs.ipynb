# Inheritance :


Inheritance means one class can use the properties and methods of another class.

- It’s like a child inheriting traits from a parent.
- Parent class → gives features.
- Child class → reuses them and can add its own.

Problem: Create an ElectricCar class that inherits from the Car class and has an additional attribute batter_size.

````python
class ElectricCar(Car):
  def __init__(self, brand, model, battery_size):
    super().__init__(brand , model)
    self.battery_size = battery_size
````

````python
car_1 = ElectricCar("Tesla" , "Model X" , "85kwh")
````

````python
car_1.full_name()
````

````python
class ElectricCar(Car):                 # → Class definition (ElectricCar is a subclass)
                                        # → Inherits from the parent class Car

    def __init__(self, brand, model, battery_size):
        # → __init__ is the Constructor method (special method to initialize objects)
        # → brand, model, battery_size are Parameters (inputs passed when creating the object)

        super().__init__(brand, model)
        # → super() is used to call the parent class (Car) constructor
        # → This reuses the initialization logic of Car

        self.battery_size = battery_size
        # → self.battery_size is an Attribute (instance variable specific to ElectricCar)
````

# When Do We Need to Use the Constructor?

A **constructor (`__init__`)** is used when we want to **initialize attributes** of an object at the time of its creation.  
This means whenever we create an object, we can directly assign values to it without writing extra lines later.

---

## Why Use a Constructor?
- To **set initial values** for attributes automatically.  
- To make sure every object starts with the required data.  
- To avoid repeating code for assigning values after object creation.

# 4 : Encapsulation:



###  What is Encapsulation?
Encapsulation means **putting data and the code that works on that data together inside one box (class)**.  
It also means **hiding the details** so that the data is safe and can only be accessed in a controlled way.

---

###  Breaking it Down
1. **Class as a Box**  
   - A class is like a box that stores both data (variables) and actions (methods).  
   - Example: `Car` class stores `brand` and `model`.

2. **Private Data**  
   - Some data is marked as *private* so nobody can directly change it from outside.  
   - Example: `self.__brand` is private.

3. **Getter/Setter Methods**  
   - To access or change private data, we use special methods (getters and setters).  
   - This ensures the data is safe and only changed in the right way.

4. **Object as a Real Thing**  
   - When you create an object (like `my_car`), it bundles the brand and model together.  
   - You then interact with the object instead of directly touching its data.

# Understanding `from abc import ABC, abstractmethod`

## What is `abc`?
- `abc` stands for **Abstract Base Classes**.
- It is a Python module that allows you to define **abstract classes** and **abstract methods**.

## Abstract Class
- An **abstract class** is a class that serves as a **blueprint** for other classes.
- It can contain:
  - **Concrete methods** (with implementation).
  - **Abstract methods** (without implementation).
- You **cannot create objects** directly from an abstract class.
- Child classes must **inherit** from the abstract class and provide implementations for all abstract methods.

## Abstract Method
- Declared using the `@abstractmethod` decorator.
- Contains only the method signature (no body).
- Forces child classes to implement this method.
- Ensures **consistency** across all subclasses.

- Declared using the `@abstractmethod` decorator inside an abstract class.
- Cannot be used directly; must be implemented in child classes.

# Concrete vs Abstract Methods

## Key Differences

| Feature            | Concrete Method                          | Abstract Method                          |
|--------------------|------------------------------------------|------------------------------------------|
| Implementation     | Has complete code inside                 | No code, only declaration                |
| Usage              | Can be used directly by child classes    | Must be implemented in child classes     |
| Purpose            | Provides ready-to-use functionality      | Enforces rules and structure             |
| Instantiation      | Available in normal classes              | Declared inside abstract classes only    |
| Example            | `database()` with `print()` statement    | `security()` with `pass`                 |


# Summary

- **Concrete methods** are fully implemented and can be directly used by subclasses.  
- **Abstract methods** act as a contract: subclasses must provide their own implementation.  
- Abstract classes can contain both concrete and abstract methods.  
- This combination allows you to:
  - Provide common functionality (via concrete methods).
  - Enforce rules and consistency (via abstract methods).  
- In Python, abstraction is achieved using the `ABC` module and the `@abstractmethod` decorator.


````python
from abc import ABC, abstractmethod

# Senior developer creates abstract BankApp class
class BankApp(ABC):

    # Concrete method - has implementation
    def database(self):
        print('connected to database')

    # Abstract method - no implementation, just declaration
    # The @abstractmethod decorator makes this method abstract
    @abstractmethod
    def security(self):
        pass
        # No code here - child classes must provide implementation


# Junior developer creates MobileApp class
class MobileApp(BankApp):

    # Only implemented mobile_login method
    def mobile_login(self):
        print("login into mobile")

    # ERROR: Did NOT implement security() method!
````

````python
from abc import ABC, abstractmethod

# Senior developer creates abstract BankApp class
class BankApp(ABC):

    # Concrete method - has implementation
    def database(self):
        print("Connected to database")

    # Abstract method - no implementation, just declaration
    @abstractmethod
    def security(self):
        pass


# Junior developer creates MobileApp class
class MobileApp(BankApp):

    def mobile_login(self):
        print("Login into mobile")

    # Real-world implementation of security
    def security(self):
        print("Security check: OTP + Fingerprint authentication")


# Junior developer creates WebApp class
class WebApp(BankApp):

    def web_login(self):
        print("Login into web")

    # Real-world implementation of security
    def security(self):
        print("Security check: Password + Captcha + 2FA")


# Junior developer creates ATMApp class
class ATMApp(BankApp):

    def atm_login(self):
        print("Login into ATM")

    # Real-world implementation of security
    def security(self):
        print("Security check: PIN + Card verification")
````

````python
# Usage
mobile = MobileApp()
mobile.database()
mobile.mobile_login()
mobile.security()
````

````python
web = WebApp()
web.database()
web.web_login()
web.security()
````

````python
atm = ATMApp()
atm.database()
atm.atm_login()
atm.security()
````

````python
# Example with our Car class
from abc import ABC, abstractmethod

# Senior developer's class (Abstract Class)
class Car(ABC):
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def full_name(self):
        return f"Brand: {self.brand} | Model: {self.model}"

    # Abstract method → must be implemented by child classes
    @abstractmethod
    def fuel_type(self):
        pass


# Junior developer's class (Concrete Class)
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    # Implementing the abstract method
    def fuel_type(self):
        return "electric"


class PetrolCar(Car):
    def __init__(self, brand, model, tank_capacity):
        super().__init__(brand, model)
        self.tank_capacity = tank_capacity

    # Implementing the abstract method
    def fuel_type(self):
        return "Petrol"
````

````python
tesla = ElectricCar("Tesla", "Model S", "100 kWh")
maruti = PetrolCar("Maruti", "Swift", "40 Liters")
````

````python
print(tesla.full_name(), "| Fuel:", tesla.fuel_type())
````

````python
print(maruti.full_name(), "| Fuel:", maruti.fuel_type())
````
