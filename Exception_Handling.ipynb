# Exception Handling in Python

##  What is Exception Handling?
Exception Handling is a mechanism in Python that allows us to **gracefully handle errors** that occur during program execution. Instead of the program crashing when an error arises, Python provides tools to **catch** and **manage** these errors, ensuring the program continues running or exits cleanly.

---

##  Why Do We Need Exception Handling?
- **Prevents program crashes**: Without handling, runtime errors stop execution abruptly.
- **Improves user experience**: Users see meaningful error messages instead of cryptic system errors.
- **Makes debugging easier**: Helps identify the cause of errors quickly.
- **Ensures reliability**: Critical applications (like banking, healthcare, etc.) must handle unexpected situations safely.
- **Supports clean-up tasks**: Ensures resources (files, connections, memory) are properly released even if errors occur.

---

##  How Does It Work?
Python uses the **`try-except` block** to handle exceptions.

### Basic Flow:
1. **`try` block** → Code that may raise an error.
2. **`except` block** → Code that runs if an error occurs.
3. **Optional `else` block** → Runs if no error occurs.
4. **Optional `finally` block** → Runs regardless of error (used for clean-up).

# Common Exceptions in Python

In Python, exceptions represent **errors that occur during program execution**. Some exceptions are very common and frequently encountered by beginners. Understanding them helps in writing robust and error-free code.

---

##  Common Exceptions

### 1. **ValueError**
- Raised when a function receives an argument of the correct type but an **inappropriate value**.
- Example scenario: Converting a non-numeric string to an integer (`int("abc")`).

---

### 2. **ZeroDivisionError**
- Raised when a number is divided by **zero**, which is mathematically undefined.
- Example scenario: `10 / 0`.

---

### 3. **FileNotFoundError**
- Raised when trying to **open or access a file** that does not exist in the specified location.
- Example scenario: `open("missing_file.txt")`.

---

### 4. **TypeError**
- Raised when an operation or function is applied to an object of an **incompatible type**.
- Example scenario: Adding a string and an integer (`"hello" + 5`).

---

##  Why Learn These?
- These exceptions are **frequent in real-world coding**.
- Knowing them helps in **anticipating errors** and writing proper `try-except` blocks.
- They form the foundation for **robust error handling** in Python.

---



## Example: Using a generic Exception handler

```python
try:
    # Code that may cause an error
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))

    result = a / b
    print("Result:", result)

except Exception as e:
    # This block catches ANY kind of error
    print("An error occurred:", e)
```

## Example of ValueError in Python

```python
try:
    # Asking user to enter a number
    a = input("Enter a number: ")

    # Trying to convert input into integer
    num = int(a)

    print("You entered:", num)

except ValueError:
    # This block runs if input is not a valid integer
    print("Error: Please enter digits only (like 5 or 10).")
```

# Difference Between Normal Exception and ValueError in Python

##  What is a "Normal Exception"?
- In Python, **Exception** is the **base class** for all built-in exceptions.
- When you write `except Exception:`, it will catch **any kind of error** (ValueError, ZeroDivisionError, TypeError, etc.).
- This is called **generic exception handling**, useful when you don’t know what type of error may occur.
- Downside: It hides the specific cause of the error, making debugging harder.

---

##  What is a "ValueError"?
- **ValueError** is a **specific type of exception** that occurs when:
  - The data type is correct, but the **value is inappropriate**.
- Example: `int("abc")` → Python expects a string of digits, but `"abc"` is not valid → raises **ValueError**.
- By catching `ValueError`, you handle only this particular case, leaving other errors to be raised normally.

---

##  Key Difference
| Aspect | Normal Exception (`Exception`) | Specific Exception (`ValueError`) |
|--------|--------------------------------|-----------------------------------|
| Scope  | Catches **all errors** | Catches **only ValueError** |
| Use    | Broad, generic handling | Targeted, precise handling |
| Debugging | Harder (cause hidden) | Easier (cause is clear) |
| Example | `except Exception:` | `except ValueError:` |

---

## Best Practice
- Use **specific exceptions** (like `ValueError`, `ZeroDivisionError`) whenever possible → clearer and safer.
- Use **generic `Exception`** only when you want to catch *any* error (e.g., logging, last-resort handling).


## Example of ZeroDivisionError in Python

```python
try:
    # Asking user for two numbers
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    # Trying to divide by the second number
    result = a / b
    print("Result:", result)

except ZeroDivisionError:
    # This block runs if the second number is zero
    print("Error: Division by zero is not allowed.")
```

## Example of FileNotFoundError in Python

```python
try:
    # Trying to open a file that does not exist
    f = open("mydata.txt", "r")
    content = f.read()
    f.close()

except FileNotFoundError:
    # This block runs if the file is missing
    print("Error: The file you are trying to open was not found.")
```

## Example of FileNotFoundError

```python
import pandas as pd

try:
    # Trying to load a CSV file that does not exist
    df = pd.read_csv("students_data.csv")
    print(df.head())

except FileNotFoundError:
    # This block runs if the file is missing
    print("Error: The file 'students_data.csv' was not found. Please check the file name or path.")
```

##  Example of TypeError in Python

```python
try:
    # Trying to add a string and an integer
    result = "Hello" + 5
    print("Result:", result)

except TypeError:
    # This block runs if incompatible types are used together
    print("❌ Error: You cannot add a string and a number directly.")
```
